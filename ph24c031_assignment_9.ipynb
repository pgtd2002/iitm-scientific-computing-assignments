## Q1

```python
def DD_checker(A):
    n=len(A)
    F=[]
    for i in range(n):
        total=sum(abs(A[i,:]))-2*abs(A[i,i])
        if total>0:
            F.append(i+1)
    if len(F)>0:
        return print(f'{A} \n Not strictly diagonally dominant on row (row number={F})')
    else:
        return print(f'{A} \n diagonally dominant')
```

```python
A=np.array([
[-6, 2, 1, 2, 1],
[3, 8, -4, 1, 0],
[-1, 1, 4, 10, 1],
[3, -4, 1, 9, 2],
[2, 0, 1, 3, 10]])

B=np.array([
[18, 3, 6, -3],
[9, 13, -5, 2],
[-3, -2, 4, 9],
[6, 0, 11, 3]])

DD_checker(A)
DD_checker(B)
```

```python
A = np.array([[-6, 2, 1, 2, 1],
              [3, 8, -4, 1, 0],
              [-1, 1, 4, 10, 1],
              [3, -4, 1, 9, 2],
              [2, 0, 1, 3, 10]], dtype=float)
B=np.array([3,4,-2,12,1], dtype=float)
DD_checker(A)
```

```python
def GS(A,B,e):
    n=len(A)
    X=np.zeros(n)
    itr=0
    err=10
    while err>e:
        X0=X.copy()
        itr+=1
        for i in range(n):
            X[i]=(B[i]-sum(A[i]*X)+A[i,i]*X[i])/A[i,i]
        err=norm(X-X0)
    return X,itr,e    
```

```python
x,y,z=GS(A,B,0.001)
print(x)
print(f'number of iteration = {y}')
```

## Q3

```python
import numpy as np

A = np.array([[2, 3, -2, 2],
              [1, 6, -4, 4],
              [4, -1, 6, 6]], dtype=float)
n = len(A)

def GE(A, n):
    for i in range(n - 1):
        for j in range(i + 1, n):
            ratio = A[j, i] / A[i, i]
            A[j, :] = A[j, :] - ratio * A[i, :]
    return A


def BS(A, n):
    X = np.zeros(n)
    B = A[:, -1]
    for i in range(n - 1, -1, -1):
        X[i] = (B[i] - sum(A[i,i+1:n]*X[i+1:n])) / A[i, i]
    return X


A = GE(A, n)
X = BS(A, n)
print(A[:,:-1]@X)
#[-2.22044605e-16  1.50000000e+00  1.25000000e+00]
```

### way 1

```python
import numpy as np

def lu_decomposition(A):
    n = A.shape[0]
    L = np.eye(n)
    U = A.copy().astype(float)
    for i in range(n):
        for j in range(i+1, n):
            L[j, i] = U[j, i] / U[i, i]
            U[j] -= L[j, i] * U[i]
    return L, U
```

```python
def forward_backward_sub(L, U, b):
    n = len(b)
    y = np.zeros(n)
    for i in range(n):
        y[i] = b[i] - np.dot(L[i, :i], y[:i])
    x = np.zeros(n)
    for i in reversed(range(n)):
        x[i] = (y[i] - np.dot(U[i, i+1:], x[i+1:])) / U[i, i]
    return x
```

```python
A=np.array([[2,3,-2,2],
            [1,6,-4,4],
            [4,-1,6,6]], dtype=float)

M=A[:,:-1]
N=A[:,-1]
L,U = lu_decomposition(M)
print(L,U)
forward_backward_sub(L, U, N)
```

```python
A=np.array([[2,3,-2,2],
            [1,6,-4,4],
            [4,-1,6,6]], dtype=float)

M=A[:,:-1]
N=A[:,-1]
L=np.tril(M)
U=M-L
err=6
X=np.zeros(len(M))
while err>0.0001:
    X0=X.copy()
    X=np.dot(inv(L),(N-np.dot(U,X)))
    err=norm(X-X0)
print(X)   # solutions
```
