## 1. Convert Radians to Degrees

```python
import numpy as np
def f(x):
    y=x*(180/np.pi)
    return(y)
x=float(input("Enter a number in radians: "))
print("number in radians: ",x)
print("value in degree:",f(x))
```

## 2. Convert Decimal to Binary

```python
def g(x):
    z=0
    k=0
    while x!=0:
        y=x%2
        x=x//2
        z=z+y*(10**k)
        k=k+1
    
    return z
x=int(input("Enter a number: "))
print("number in decimal: ",x)
print('binary number is-',g(x))
```

## 3. Perform Arithmetic Operations

```python
def calculate(a, b, operator):
    if operator == "+":
        return a + b
    elif operator == "-":
        return a - b
    elif operator == "*":
        return a * b
    elif operator == "/":
        return a / b if b != 0 else "Division by zero is not allowed"
    else:
        return "Invalid operator"

# Example usage
num1, num2 = float(input('1st number')), float(input('2nd number'))
operator = input('symbol of operation')
print(num1,operator,num2,'=',calculate(num1, num2, operator))
```

## 4. Simple Interest Calculation

```python
def simple_interest(principal, time, rate):
    return (principal * time * rate) / 100

# Example usage
P, T, R = float(input('principal')), float(input('time(yearly)')), float(input('rate'))
print(f"Simple Interest = {simple_interest(P, T, R)}")
```

## 5. Double Each Character in a String

```python
def double_characters(string):
    return "".join([char * 2 for char in string])

# Example usage
input_string = "Python"
print(input_string)
print(f"Doubled string: {double_characters(input_string)}")
```

## 6. Factorial of a Number

```python
def factorial(x):
    n=1
    for i in range(x):
        n=n*(i+1)
    return (n)
z=int(input("Enter a number: "))
print(z,'! =',factorial(z))
    
```

## 7. Fibonacci Number Using Recursion

```python
def fibonacci(n):
    if n <= 0:
        return "Invalid input"
    elif n == 1:
        return 0
    elif n == 2:
        return 1
    else:
        return fibonacci(n - 1) + fibonacci(n - 2)

# Example usage
nth_fib = int(input("Enter a number: "))
print(f"The {nth_fib}-th Fibonacci number = {fibonacci(nth_fib)}")
```
