### 1. Plot Sin(x)+Cos(y)*Sin(y)*Cos(x)*Cos(x) for x,y range −2π to 2π. Use a 2D grid with dimensions 128×128.

```python
from mpl_toolkits import mplot3d
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(-2*np.pi, np.pi, 128)
y=np.linspace(-2*np.pi, np.pi, 128)
x,y=np.meshgrid(x,y)
z=np.sin(x)+(np.cos(y)*np.sin(y)*np.cos(x)*np.cos(x))
ax=plt.axes(projection='3d')
ax.plot_surface(x, y, z,cmap='viridis', alpha=1)
plt.show()
plt.imshow(z,extent=[-2*np.pi, 2*np.pi, -2*np.pi, 2*np.pi],cmap='viridis')
plt.colorbar(label='Value of function')
plt.show()
```

## 2. Plot a 2D Gaussian function for x = −5 to +5 and y = −5 to +5.

```python
x=np.linspace(-5,5,400)
y=np.linspace(-5,5,400)
x,y=np.meshgrid(x,y)
z=np.exp(-x**2-y**2)
ax=plt.axes(projection='3d')
ax.plot_surface(x, y, z,cmap='viridis', alpha=1)
plt.show()
plt.imshow(z,extent=[-5,5,-5,5],cmap='viridis')
plt.colorbar(label='Value of function')
plt.show()
```

## 3. Evaluation of Sin(x) within a given error limit by adding up the series: Use power series expansion in order to evaluate Sin(x) for a given x. Truncate the series when the value reaches within the accuracy of allowed error e (User defined error limit) set by you.

```python
import math as mt
import numpy as np
err=10**-5
def f(x):
    term=0.1
    i=0
    z=0
    
    while abs(term)>err:
        i=i+1
        term=(x**(2*i-1))*((-1)**(i-1))/mt.factorial(2*i-1)
        z=z+term
    return z
c=float(np.pi/3)
print(f"calculated using power series-{f(c)}(with error-{err})")
print(f"default numpy value-{np.sin(np.pi/3)}")
```

```python
import math as mt
import numpy as np
def e(x,N):
    z=0
    for i in range(N+1):
        term=((-x)**i)/(mt.factorial(i))
        z=z+term
    return z
X=[0,10,50,100]

def error(x,N):
    z=abs((e(x,N)-np.exp(-x)))/np.exp(-x)
    return np.log(z)

def ploting(x):
    N=np.arange(0,1000,10)
    N=np.delete(N,0)
    ERR=[]
    for i in N:
        ERR.append(error(x,i))
    plt.loglog(N,ERR)
    plt.title(f"Error for x={x}")
    #plt.yscale('log')
    plt.xlabel("Number of terms in sum")
    plt.ylabel("Error")
    return plt.show()

    
for x in X:
    print(f"{ploting(x)}")
    
```
